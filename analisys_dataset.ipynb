# Dataset Analysis

```python
import pandas as pd

df = pd.read_csv('customer-dataset.csv')
df
```

```python
categories = set(df['category'])
print(categories)
print(len(categories))
```

```python
df.groupby(["category"]).size().reset_index(name="count")
```

```python
df.groupby(["category", "intent"]).size().reset_index(name="count")
```

```python
categories_selected = ['ACCOUNT', 'ORDER', 'REFUND']
intents_selected = ['create_account', 'delete_account', 'edit_account',
                    'cancel_order', 'change_order', 'place_order',
                    'check_refund_policy', 'get_refund', 'track_refund'
                   ]
ds_filtered = df[
    df['category'].isin(categories_selected) &
    df['intent'].isin(intents_selected)
]
ds_filtered
```

```python
import plotly.express as px

fig = px.bar(
    ds_filtered['intent'].value_counts().reset_index(),
    x='intent',
    y='count',
    title='Number of Records per Intent',
    text_auto=True
)

fig.show()
```

```python
ds_filtered.info()
```

```python
sampled = ds_filtered.groupby('intent', group_keys=False).sample(n=6, random_state=42)

train_df = sampled.groupby('intent', group_keys=False).head(3)
print(f"Trainig Dataset Length: {len(train_df)}")
train_df
```

```python
test_df = sampled.drop(train_df.index)
print(f"Test Dataset Length: {len(test_df)}")
test_df
```
